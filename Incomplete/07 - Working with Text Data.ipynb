## Common String Methods
- A `Series` has a special `str` attribute that exposes an object with string methods.
- Access the `str` attribute, then invoke the string method on the nested object.
- Most method names will match their Python method equivalents (`upper`, `lower`, `title`, etc).

```python
city = pd.read_csv('chicago.csv').dropna(how='all').astype({'Department': 'category'})
city.head()
```

```python
city['Position Title']
```

```python
type(city['Position Title'])
```

```python
city['Position Title'].str.lower()
city['Position Title'].str.upper()
city['Position Title'].str.title()
city['Position Title'].str.len()
```

```python
city['Position Title'].str.title().str.len()
```

```python
city['Position Title'].str.strip() # removes whitespaces
city['Position Title'].str.lstrip() # remove begining whitespace
city['Position Title'].str.rstrip() # remove end whitespace
```

```python
city['Position Title'].str.count('A') # counts A in the text
```

```python
city['Department']
```

```python
city['Department'].str.replace('MGMNT', 'Management')
```

- The `str` attribute exposes an object that supports square bracket syntax for indexing.
- You can provide values, slices, or negative numbers.

```python
city['Department'].str[0:6]
```

```python
city['Department'].str[::-1]
```

## Filtering with String Methods
- The `str.contains` method checks whether a substring exists anywhere in the string.
- The `str.startswith` method checks whether a substring exists at the start of the string.
- The `str.endswith` method checks whether a substring exists at the end of the string.

```python
city = pd.read_csv('chicago.csv').dropna(how='all').astype({'Department': 'category'})
city.head()
```

```python
water_workers = city['Position Title'].str.lower().str.contains('water')
city[water_workers]
```

```python
water_workers = city['Position Title'].str.upper().str.contains('WATER')
city[water_workers]
```

```python
water_workers = city['Position Title'].str.contains('water', case=False)
city[water_workers]
```

```python
start_with_civil = city['Position Title'].str.lower().str.startswith('civil')
city[start_with_civil]
```

```python
end_with_iv = city['Position Title'].str.lower().str.strip().str.endswith('iv')
city[end_with_iv]
```

## String Methods on Index and Columns
- Use the `index` and `columns` attributes to access the `DataFrame` index/column labels.
- These objects support string methods via their own `str` attribute.

```python
city = pd.read_csv('chicago.csv', index_col='Name').dropna(how='all').astype({'Department': 'category'})
city.head()
```

```python
city.columns
```

```python
city.columns.str.upper()
```

```python
city.columns = city.columns.str.upper()
```

```python
city.head()
```

## The str.split Method
- The `str.split` method splits a string by the occurrence of a delimiter. Pandas returns a `Series` of lists.
- Use the `str.get` method to access a nested list element by its index position.

```python
city = pd.read_csv('chicago.csv').dropna(how='all').astype({'Department': 'category'})
city
```

```python
city['Position Title'].str.split(" ").str.get(0).value_counts()
```

- Let's find the most common first name among the employees.

```python
city['Name'].str.split(', ').str.get(1).str.strip().str.split(' ').str.get(0).value_counts()
```

## The expand and n Parameters of the str.split Method
- The `expand` parameter returns a `DataFrame` instead of a `Series` of lists.
- The `n` parameter limits the number of splits.

```python
city = pd.read_csv('chicago.csv').dropna(how='all').astype({'Department': 'category'})
city.head()
```

```python
city[['last Name', 'First Name']] = city['Name'].str.title().str.split(', ', expand=True)
```

```python
city.head()
```

```python
city[['Primary Title', 'Secondary Title']] = city['Position Title'].str.split(' ', expand=True, n=1)
```

```python
city.head()
```

## The explode Method
- The `explode` method transforms a `Series` of lists into a column with a separate row for each list element.
- Pandas preserves the original index. Use `reset_index` with `drop=True` to generate a new sequential index.

```python
emp_skills = pd.read_csv('employee_skills.csv')
emp_skills
```

```python
emp_skills['Skills'] = emp_skills['Skills'].str.split('|')
```

```python
emp_skills.head()
```

```python
emp_skills.explode('Skills')
```

```python
emp_skills.explode('Skills').reset_index()
```

```python
emp_skills.explode('Skills').reset_index(drop=True)
```
